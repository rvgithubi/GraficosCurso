```python
# prompt: read a csv file with pandas using utf-8 code

import pandas as pd
df = pd.read_csv('uni.csv', encoding='utf-8')
#df = pd.read_csv('uni.csv', encoding='utf-8',sep=";") #casos que CSV separa ; |
print(df.head())
```

```python
# prompt: necesito visualizar los datos de las columnas del dataframe df

print(df.head())
```

```python
# prompt: requiero agrupar el dataframe df por la columna ANIO y contar los registros por cada ANIO

print(df.groupby('ANIO').size())
```

```python
# prompt: requiero crear un nuevo dataframe con doble agrupacion ANIO y DOMICILIO_DEPA para saber la cantidad de registros por ANIO y DOMICILIO_DEPA

df_agrupado = df.groupby(['ANIO', 'DOMICILIO_DEPA']).size().reset_index(name='cantidad')
print(df_agrupado)
```

```python
# prompt: requiero filtrar del dataframe df_agrupado solo los registros que DOMICILIO_DEPA sea AMAZONAS

df_filtrado_amazonas = df_agrupado[df_agrupado['DOMICILIO_DEPA'] == 'AREQUIPA']
df_filtrado_amazonas
```

```python
# prompt: crear un grafico en barras para mostrar el dataframe df_filtrado_amazonas

import matplotlib.pyplot as plt

df_filtrado_amazonas = df_agrupado[df_agrupado['DOMICILIO_DEPA'] == 'APURÍMAC']
plt.figure(figsize=(10, 6))
plt.bar(df_filtrado_amazonas['ANIO'], df_filtrado_amazonas['cantidad'])
plt.xlabel('Año')
plt.ylabel('Cantidad de matrículas')
plt.title('Matrículas en Amazonas por Año')
plt.xticks(df_filtrado_amazonas['ANIO']) # Ensure all years are shown as ticks
plt.grid(axis='y')
plt.show()
```
