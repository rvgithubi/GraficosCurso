```python
import yfinance as yf  #Yahoo Finance
```

```python
symbol="NVDA"
stock=yf.Ticker(symbol)
```

```python
try:
  price=stock.history(period="1d")["Close"].iloc[-1]  #Precio cierre
  print(f" Ok {symbol}: {price:.2f}")
except IndexError:
  print(f"Error retrieving price for {symbol}")
```

```python
symbol="BTC-USD"
stock=yf.Ticker(symbol)
try:
  price=stock.history(period="1d")["Close"].iloc[-1]
  print(f"{symbol}: {price:.2f}")
except IndexError:
  print(f"Error retrieving price for {symbol}")
```

```python
symbol="ADA-USD"
stock=yf.Ticker(symbol)
try:
  price=stock.history(period="1d")["Close"].iloc[-1]
  print(f"{symbol}: {price:.3f}")
except IndexError:
  print(f"Error retrieving price for {symbol}")
```

```python
try:
  btc_monthly_prices = stock.history(period="1mo")["Close"].tolist()
  print(f"{symbol} last month daily prices: {btc_monthly_prices}")
except IndexError:
  print(f"Error retrieving monthly prices for {symbol}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(btc_monthly_prices)
plt.title(f'{symbol} Last Month Daily Prices')
plt.xlabel('Days')
plt.ylabel('Price')
plt.grid(True)
plt.show()
```

```python
try:
  btc_3month_prices = stock.history(period="3mo")["Close"].tolist()
  print(f"{symbol} last 3 months daily prices: {btc_3month_prices}")
except IndexError:
  print(f"Error retrieving 3 month prices for {symbol}")
```

```python
plt.figure(figsize=(12, 6))
plt.plot(btc_3month_prices)
plt.title(f'{symbol} Last 3 Months Daily Prices')
plt.xlabel('Days')
plt.ylabel('Price')
plt.grid(axis='y')
plt.show()
```
