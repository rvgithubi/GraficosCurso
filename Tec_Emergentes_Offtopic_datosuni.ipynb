```python
import pandas as pd
df = pd.read_csv('uni.csv', encoding='utf-8') #lectura del csv, paginacion/codificación
#df = pd.read_csv('uni.csv', encoding='utf-8',sep=";") #casos que CSV separa ; |
print(df.head())
```

```python
# prompt: necesito visualizar los datos de las columnas del dataframe df

print(df.head())
```

```python
print(df.groupby('ANIO').size())
```

```python
print(df.groupby(['SEXO','ANIO']).size())
```

```python
print(df.groupby(['ANIO', 'SEXO']).size())
```

```python
df_anio_sezo=df.groupby(['ANIO', 'SEXO']).size()
print("Año |  Sexo | Cantidad" )
for index, value in df_anio_sezo.items():
    print(f"{index[0]} |  {index[1]} |  {value}")
```

```python
df_agrupado = df.groupby(['ANIO', 'DOMICILIO_DEPA']).size().reset_index(name='cantidad')

#for index, value in df_agrupado.items():
#    print(f"{index[0]} |  {index[1]} |  {value}")
print(df_agrupado)
```

```python
import matplotlib.pyplot as plt  #gráficos

df_filtrado_amazonas = df_agrupado[df_agrupado['DOMICILIO_DEPA'] == 'CAJAMARCA']
plt.figure(figsize=(10, 6))
plt.bar(df_filtrado_amazonas['ANIO'], df_filtrado_amazonas['cantidad'])
plt.xlabel('Año')
plt.ylabel('Cantidad de matrículas')
plt.title('Matrículas en Cajamaca por Año')
plt.xticks(df_filtrado_amazonas['ANIO']) # Ensure all years are shown as ticks
plt.grid(axis='y')
plt.show()
```
